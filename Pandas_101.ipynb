###การบ้าน

````
this_data = data_covid[['sex','age','province_of_onset','risk']]

# ใช้ fillna แทนที่ missing ใน sex / province ด้วยค่าที่เหมาะสม
this_data = this_data.fillna(value={'sex':'ไม่ทราบ', 'province_of_onset':'ไม่ทราบ'})

# ลบ record ที่ age เป็นค่าเพี้ยน
this_data = this_data[this_data['age'] != '55/664']

# ลบ record ที่ age เป็น missing
this_data = this_data.dropna(subset=['age'])

# ---------- ข้อ 1: อายุเฉลี่ยของผู้หญิง/ผู้ชาย ทั้งหมด ----------
female = this_data[this_data['sex']=='หญิง']
male   = this_data[this_data['sex']=='ชาย']

sum_age = 0
N = 0
for a in female['age']:
    a = float(a)
    if a > 0:
        sum_age += a
        N += 1
print(f'อายุเฉลี่ย ของ ผู้ป่วยหญิง (ทั้งหมด): {sum_age/N}')

sum_age = 0
N = 0
for a in male['age']:
    a = float(a)
    if a > 0:
        sum_age += a
        N += 1
print(f'อายุเฉลี่ย ของ ผู้ป่วยชาย (ทั้งหมด): {sum_age/N}')

print('-'*30)

# ---------- ข้อ 2: อายุเฉลี่ย เฉพาะขอนแก่น ----------
female_KK = female[female['province_of_onset']=='ขอนแก่น']
male_KK   = male[male['province_of_onset']=='ขอนแก่น']

sum_age = 0
N = 0
for a in female_KK['age']:
    a = float(a)
    if a > 0:
        sum_age += a
        N += 1
print(f'อายุเฉลี่ย ของ ผู้ป่วยหญิง ในจังหวัดขอนแก่น: {sum_age/N}')

sum_age = 0
N = 0
for a in male_KK['age']:
    a = float(a)
    if a > 0:
        sum_age += a
        N += 1
print(f'อายุเฉลี่ย ของ ผู้ป่วยชาย ในจังหวัดขอนแก่น: {sum_age/N}')

print('-'*30)

# ---------- ข้อ 3: จำนวนผู้ป่วยคนต่างชาติเดินทางมาจากต่างประเทศ ----------
foreign_travelers_count = this_data[this_data['risk'] == 'คนต่างชาติเดินทางมาจากต่างประเทศ'].shape[0]
print(f'จำนวนผู้ป่วยที่เป็น "คนต่างชาติเดินทางมาจากต่างประเทศ": {foreign_travelers_count} ราย')
````
